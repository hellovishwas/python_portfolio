## "is" and "is not"

```python
a="10" #For single data id is same if value stored is same
b="10"
print(id(a))
print(id(b))
a is b
```

```python
a=[1,2,3]#For collection data id is different no matter  value stored is same
b=[1,2,3]
print(id(a))
print(id(b))
a is b
```
