```python
for i in range (1,100):
    if i%2==0:
        print("vishwas")
    else:
        print(i)
```

```python
for i in range (1,100):
    if i%2==0:
        continue
    else:
        print(i)
```

```python
for i in range (1,100):
    if i%2==0:
        break
    else:
        print(i)
```

```python
for i in range (1,100):
    if i%2==0:
        pass
    else:
        print(i)
```

```python
for i in range(100):
    if i == 34:
        break  # exit the loop right now
    print(i)
```
