### Set functions
- There are many Python functions that are useful to manipulate sets
- Those include:
    - add()
    - remove()
    - update()

```python
# add() adds one element to set
lstset = {1, 10, 2}
s4  = {4,5, 6}

lstset.add(5)
print(lstset)
```

```python
# remove() deletes certain element in set
s4 = {2, 4, 5}
s4.remove(5)
print(s4)
```

```python
# update() adds more than one element to set
s5 = {2, 7, 3}
s5.update([2, 4,1,34,45,6, 5])
print(s5)
```

The update() method updates the current set, by adding items from another set (or any other iterable).
If an item is present in both sets, only one appearance of this item will be present in the updated set.





```python
a = {1,2,3}
b = {2,3,4}
a.update(b)
print(a)
```

Use |= as a shortcut instead of update():



```python
a = {1,2,3}
b = {2,3,4}
a|=b
print(a)
```

## 4. Dictionary
- Dictionary elements comprise each key-value pair
- Dictionary does not have order; dictionary elements can be accessed via keys

```python
# When creating dictionaries, value for each key should be addressed
d1 = {1: 'John', 2: 'Jane', 3: 'Jackson'}
print(type(d1))
print(d1[1])
print(d1[2])

```

```python
d2 = {'a': 1, 'b': 2, 'c': 3}
print(d2)

d2['d'] = 4                        # adding element
print(d2)

del d2['a']                        # deleting element
print(d2)
```

### Dictionary functions
- items()
- keys()
- values()

```python
d3 = {'AL': 'Alabama', 'AK': 'Alaska', 'AZ': 'Arizona', 'AR': 'Arkansas', 'CA': 'California'}
```

```python
# items() returns each item in tuple
d3_items = d3.items()
print(d3_items)
```

```python
# keys() return all keys in tuple
d3_keys = d3.keys()
print(d3_keys)
```

```python
# values() return all values in list
d3_values = d3.values()
print(d3_values)
```
